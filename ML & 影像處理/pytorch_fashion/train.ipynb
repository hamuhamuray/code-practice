```python
import torch
import torchvision.datasets
import torchvision.transforms
import torch.utils.data.dataloader
```

```python
train_data = torchvision.datasets.FashionMNIST(root = "./", train = True, download = True, transform = torchvision.transforms.ToTensor())
test_data = torchvision.datasets.FashionMNIST(root = "./", train = False, download = True, transform = torchvision.transforms.ToTensor())
# ToTensor()：將圖片轉成張量，並map至[0, 1]
```

```python
train_loader = torch.utils.data.DataLoader(dataset = train_data, batch_size = 64, shuffle = True)

for batch in train_loader:
    print(type(batch))
    break
```

```python
class FashionModel(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = torch.nn.Linear(784, 128)
        self.relu = torch.nn.ReLU()
        self.output = torch.nn.Linear(128, 10)
        self.flatten = torch.nn.Flatten()
        self.mymodel = torch.nn.Sequential(
            self.flatten,
            self.linear,
            self.relu,
            self.output
        )

    def forward(self, x):
        return self.mymodel(x)

```

```python
images, labels = next(iter(train_loader))
model = FashionModel()
pred = model(images)
print(images.shape)
print(pred.shape)
print(pred[0])
print(pred[0].sum())
```

```python
prediction = torch.argmax(pred, dim = 1)
print(prediction.shape)
print(prediction[:10])
print(labels[:10])
```

```python
loss_fn = torch.nn.CrossEntropyLoss()
loss = loss_fn(pred, labels)
print(type(loss))
print(type(loss.item()))
print(loss)
print(loss.item())
```

```python
device = "cuda"
model = FashionModel().to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
loss_fn = torch.nn.CrossEntropyLoss()
for epoch in range(5):
    total_loss = 0
    for images, labels in train_loader:
        images = images.to(device)
        labels = labels.to(device)
        pred = model(images)
        loss = loss_fn(pred, labels)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    avg_loss  = total_loss / len(train_loader)
    print(f"Epoch {epoch + 1}: Loss = {avg_loss:.3f}")
```

```python
test_loader = torch.utils.data.DataLoader(dataset = test_data, batch_size = 64)

model.eval()
with torch.no_grad():
    correct = 0
    label_count = 0
    for images, labels in test_loader:
        images = images.to(device)
        labels = labels.to(device)
        pred = model(images)
        prediction = torch.argmax(pred, dim=1)
        correct += (prediction == labels).sum().item()
        label_count += len(labels)
    accuracy = correct / label_count
    print(accuracy)
```

```python
class FashionConvModel(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.conv2d_1 = torch.nn.Conv2d(in_channels = 1, out_channels = 16, kernel_size = 3, padding = 1)
        self.relu = torch.nn.ReLU()
        self.maxpool2d = torch.nn.MaxPool2d(kernel_size = 2)
        self.conv2d_2 = torch.nn.Conv2d(in_channels = 16, out_channels = 32, kernel_size = 3, padding = 1)
        self.flatten = torch.nn.Flatten()
        self.linear = torch.nn.Linear(in_features = 1568, out_features = 128)
        self.output = torch.nn.Linear(in_features = 128, out_features = 10)
        self.model = torch.nn.Sequential(
            self.conv2d_1,
            self.relu,
            self.maxpool2d,
            self.conv2d_2,
            self.relu,
            self.maxpool2d,
            self.flatten,
            self.linear,
            self.relu,
            self.output
        )

    def forward(self, x):
        return self.model(x)
```

```python
device = "cuda"
images, labels = next(iter(train_loader))
images = images.to(device)

model = FashionConvModel().to(device)
pred = model(images)

print(pred.shape)
```

```python
device = "cuda"
model = FashionConvModel().to(device)
optimizer = torch.optim.Adam(params = model.parameters(), lr = 0.001)
loss_fn = torch.nn.CrossEntropyLoss()
for epoch in range(5):
    total_loss = 0
    for images, labels in train_loader:
        images = images.to(device)
        labels = labels.to(device)
        pred = model(images)
        loss = loss_fn(pred, labels)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    print(f"Epoch {epoch + 1}: Average loss: {(total_loss / len(train_loader)):.3f}")


```

```python
test_loader = torch.utils.data.DataLoader(dataset = test_data, batch_size = 64)

model.eval()
with torch.no_grad():
    correct = 0
    cases = 0
    for images, labels in test_loader:
        images = images.to(device)
        labels = labels.to(device)
        pred = model(images)
        prediction = torch.argmax(pred, dim = 1)
        correct += (prediction == labels).sum()
        cases += len(labels)
    accuracy = correct / cases
    print(f"{accuracy:.3f}")

```

```python
new_model = FashionConvModel()
state_dict = torch.load("fashion_cnn.pth")
new_model.load_state_dict(state_dict)
```

```python
confusion_matrix = torch.zeros(10, 10).to(device)
new_model.eval()
new_model = new_model.to(device)
num_classes = 10
with torch.no_grad():
    correct = 0
    cases = 0
    for images, labels in test_loader:
        images = images.to(device)
        labels = labels.to(device)
        pred = new_model(images)
        prediction = torch.argmax(pred, dim = 1)
        correct += (prediction == labels).sum()
        cases += len(labels)
        # 更新confusion matrix
        index = labels * num_classes + prediction
        confusion_matrix += torch.bincount(index, minlength = num_classes * num_classes).reshape(num_classes, num_classes)
    accuracy = correct / cases
    print(f"{accuracy:.3f}")

print(confusion_matrix.cpu().int())
```
